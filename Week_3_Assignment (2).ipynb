```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]

plt.scatter(study_hours, exam_scores, color="blue", marker="o")

plt.title("Study Time vs. Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.show()
```

As study time increases, exam scores also increase for the most part.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
city_a = [68, 70, 72, 71, 74, 76, 75]
city_b = [62, 64, 67, 66, 69, 71, 70]

plt.plot(days, city_a, marker="o", linestyle="-", label="City A")
plt.plot(days, city_b, marker="s", linestyle="--", label="City B")

plt.title("Temperature Over One Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.legend()

plt.show()
```

City A was warmer during the week because its temperature was higher than City B's on every day.

```python
offenses = [
    "Aggravated Assault",
    "Burglary",
    "Larceny",
    "Motor Vehicle Theft",
    "Murder",
    "Rape",
    "Robbery"
]

crime_rates = [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]

colors = ["blue", "green", "orange", "red", "purple", "pink", "brown"]

plt.figure(figsize=(10, 6))

plt.bar(offenses, crime_rates, color=colors)

plt.title("Crime Rates per 100,000 People in 2023")
plt.xlabel("Offense")
plt.ylabel("Offenses per 100,000 People")

plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
offenses = [
    "Aggravated Assault",
    "Burglary",
    "Larceny",
    "Motor Vehicle Theft",
    "Murder",
    "Rape",
    "Robbery"
]

changes = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]

colors = ["blue", "green", "orange", "red", "purple", "pink", "brown"]

plt.figure(figsize=(10, 6))

plt.barh(offenses, changes, color=colors)

plt.title("Change in Crime Rates from 2022 to 2023")
plt.xlabel("Change in Offense Rate (%)")
plt.ylabel("Offense")

plt.axvline(0, color="black")
plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

Task 3A only shows the size of each crime rate in 2023, while this graph makes it easier to see which crimes increased or decreased and by how much.

```python
offenses = [
    "Aggravated Assault",
    "Burglary",
    "Larceny",
    "Motor Vehicle Theft",
    "Murder",
    "Rape",
    "Robbery"
]

changes = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]

# Red = decrease, green = increase
colors = ["red" if change < 0 else "green" for change in changes]

plt.figure(figsize=(10, 6))

plt.barh(offenses, changes, color=colors)

plt.title("Change in Crime Rates from 2022 to 2023")
plt.xlabel("Change in Offense Rate (%)")
plt.ylabel("Offense")

plt.axvline(0, color="black")
plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

x = np.arange(len(companies))
width = 0.25

plt.figure(figsize=(9, 6))

plt.bar(x - width, product1, width, label="Product 1")
plt.bar(x, product2, width, label="Product 2")
plt.bar(x + width, product3, width, label="Product 3")

plt.title("Product Sales by Company")
plt.xlabel("Company")
plt.ylabel("Sales")

plt.xticks(x, companies)
plt.legend()

plt.tight_layout()
plt.show()
```
